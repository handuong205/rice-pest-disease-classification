```python
import os

dataset_path = r"D:\KÌ 7\RiceDataset"

for cls in os.listdir(dataset_path):
    class_path = os.path.join(dataset_path, cls)

    if os.path.isdir(class_path):
        image_count = 0

        for root, dirs, files in os.walk(class_path):
            image_count += len([
                f for f in files
                if f.lower().endswith(('.jpg', '.jpeg', '.png', '.bmp', '.webp'))
            ])

        print(f"{cls}: {image_count}")
```

```python
import os
from collections import defaultdict

dataset_path = r"D:\KÌ 7\RiceDataset"

class_counts = defaultdict(int)

for source in os.listdir(dataset_path):

    source_path = os.path.join(dataset_path, source)

    if not os.path.isdir(source_path):
        continue

    for cls in os.listdir(source_path):

        class_path = os.path.join(source_path, cls)

        if not os.path.isdir(class_path):
            continue

        count = 0

        for root, dirs, files in os.walk(class_path):
            count += len([
                f for f in files
                if f.lower().endswith(
                    ('.jpg', '.jpeg', '.png', '.bmp', '.webp')
                )
            ])

        class_counts[cls] += count

for cls, count in sorted(class_counts.items()):
    print(f"{cls}: {count}")
```
